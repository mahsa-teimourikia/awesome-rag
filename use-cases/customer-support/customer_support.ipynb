# Customer-support RAG assistant

This use case combines permission filtering, query rewriting, reranking, citations, and human escalation. The system drafts answers; it does not perform refunds or account changes.

## Workflow

```mermaid
flowchart LR
 Q[Question] --> S{Sensitive?}
 S -->|yes| H[Human review]
 S -->|no| P[Permission filter]
 P --> R[Retrieve + rerank]
 R --> C{Evidence?}
 C -->|yes| A[Cited draft]
 C -->|no| E[Clarify or escalate]
```

````python
import importlib.util
from pathlib import Path
spec = importlib.util.spec_from_file_location('customer_support', Path('use-cases/customer-support/app.py'))
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)

print(module.support_answer('acme', 'How do I rotate an API key?'))
print('--- sensitive request ---')
print(module.support_answer('acme', 'Please refund my last invoice'))
````

The implementation keeps authorization before candidate construction and uses an explicit human-review route for sensitive intents. In production, use a real identity provider, document ACLs, audit logs, calibrated classifiers, and a support queue rather than keyword matching alone.
