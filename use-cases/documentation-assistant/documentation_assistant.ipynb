# Documentation assistant capstone

This project combines the beginner RAG modules into a documentation assistant. It answers only from versioned fixture docs, returns citations, and abstains outside the corpus.

## Design

The system is intentionally lexical and local. That makes the policy and provenance visible before we introduce embeddings, hybrid retrieval, permissions, freshness, and observability.

```text
FLOW (read top to bottom)

+--------------------+
| Question           |
+--------------------+
          |
          v
+--------------------+
| Retrieve           |
+--------------------+

Supporting paths:
  [Docs] --feeds--> [Retrieve]
  [Evidence policy] --supported--> [Cited answer]
  [Evidence policy] --unsupported--> [Abstain]
```
