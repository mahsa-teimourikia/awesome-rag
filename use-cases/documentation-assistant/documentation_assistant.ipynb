# Documentation assistant capstone

This project combines the beginner RAG modules into a documentation assistant. It answers only from versioned fixture docs, returns citations, and abstains outside the corpus.

## Design

The system is intentionally lexical and local. That makes the policy and provenance visible before we introduce embeddings, hybrid retrieval, permissions, freshness, and observability.

```mermaid
flowchart LR
 D[Docs] --> R[Retrieve]
 Q[Question] --> R --> P{Evidence policy}
 P -->|supported| A[Cited answer]
 P -->|unsupported| X[Abstain]
```

````python
from pathlib import Path
from examples.beginner.citations import answer_with_citations, render_markdown
from examples.beginner.first_local_rag import load_chunks

docs = Path('use-cases/documentation-assistant/docs')
chunks = load_chunks(docs)
result = answer_with_citations('How do I rotate an API key?', chunks)
print(render_markdown(result))
````

````python
unsupported = answer_with_citations('How do I order lunch?', chunks)
print(render_markdown(unsupported))
assert unsupported.abstained
````
