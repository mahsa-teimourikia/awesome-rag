# RAG from scratch

**Scenario:** NovaTech is building an Enterprise Knowledge Assistant for HR, finance, IT, project, and vendor knowledge.

Build NovaTech Enterprise Knowledge Assistant by hand: load documents, chunk them, retrieve evidence, generate cited answers, and compare against an unsupported answer.

```mermaid
flowchart LR
  Q[User question] --> R[Retrieve evidence]
  R --> V[Verify / rerank]
  V --> A[Answer with citations or abstain]
  A --> E[Evaluate failure]
```

## Learning pattern

1. Start from a deliberately simple baseline.
2. Break it with a realistic failure case.
3. Improve one component.
4. Measure the change and write down the architecture lesson.

````python
from pathlib import Path
import sys, json
ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    ROOT = ROOT.parent.parent
sys.path.insert(0, str(ROOT))
````

````python
from src.enterprise_rag.corpus import load_corpus, chunk_documents
from src.enterprise_rag.retrieval import retrieve
from src.enterprise_rag.generation import naive_answer, answer_with_citations
docs = load_corpus(ROOT / 'data/enterprise')
chunks = chunk_documents(docs)
query='What increased by 14% in Q2 2025?'
hits=retrieve(query, chunks, method='bm25', top_k=5)
print('Naive:', naive_answer(query))
print(json.dumps(answer_with_citations(query, hits), indent=2))
````
