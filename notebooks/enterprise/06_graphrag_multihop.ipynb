### When not to use GraphRAG

GraphRAG adds modeling and operational cost. For isolated definitions, exact identifiers, or short factual lookups, hybrid text retrieval may be simpler and better. Use graph retrieval when relationships, paths, neighborhoods, or corpus-level synthesis are the actual problem.
