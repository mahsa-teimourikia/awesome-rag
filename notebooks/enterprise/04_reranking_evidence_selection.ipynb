# Reranking and evidence selection

**Scenario:** NovaTech is building an Enterprise Knowledge Assistant for HR, finance, IT, project, and vendor knowledge.

Retrieve a noisy top-30 candidate set, then rerank/evidence-select the snippets that should reach the generator.

```mermaid
flowchart LR
  Q[User question] --> R[Retrieve evidence]
  R --> V[Verify / rerank]
  V --> A[Answer with citations or abstain]
  A --> E[Evaluate failure]
```

## Learning pattern

1. Start from a deliberately simple baseline.
2. Break it with a realistic failure case.
3. Improve one component.
4. Measure the change and write down the architecture lesson.

````python
from pathlib import Path
import sys, json
ROOT = Path.cwd()
if not (ROOT / 'data').exists():
    ROOT = ROOT.parent.parent
sys.path.insert(0, str(ROOT))
````

````python
from src.enterprise_rag.corpus import load_corpus, chunk_documents
from src.enterprise_rag.retrieval import hybrid_retrieve, tokenize
chunks=chunk_documents(load_corpus(ROOT/'data/enterprise'))
query='Does the Atlas supplier need R-17 certification?'
candidates=hybrid_retrieve(query, chunks, top_k=30)
def rerank(q, hits):
    q_terms=set(tokenize(q)) | {'acme','regulation','r-17','vectordb-x'}
    return sorted(hits, key=lambda h: len(q_terms & set(tokenize(h[0].text))), reverse=True)
print([(c.document_id, c.metadata['section']) for c,_ in rerank(query,candidates)[:5]])
````
